```python
%sql
CREATE OR REPLACE TABLE employees (
    id INT,
    name STRING,
    salary DOUBLE
)
```

```python
%sql
UPDATE employees 
SET salary = salary + 100
WHERE name = "Adam";

UPDATE employees 
SET salary = salary + 100
WHERE name = "Anna";
```
