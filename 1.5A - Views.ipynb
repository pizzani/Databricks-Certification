Stored Views

```python
%sql
CREATE VIEW view_apple_phones
AS  SELECT * 
    FROM smartphones 
    WHERE brand = 'Apple';
```

```python
%sql
SELECT * FROM view_apple_phones;
```

```python
%sql
show tables
```

Temporary View

```python
%sql
CREATE TEMP VIEW temp_view_phones_brands
AS  SELECT DISTINCT brand
    FROM smartphones;

SELECT * FROM temp_view_phones_brands;
```

```python
%sql
SHOW TABLES;
```

Global Temporary Views

```python
%sql
CREATE GLOBAL TEMP VIEW global_temp_view_latest_phones
AS SELECT * FROM smartphones
    WHERE year > 2020
    ORDER BY year DESC;
```

```python
%sql
CREATE OR REPLACE TEMP VIEW temp_view_latest_phones AS
SELECT * FROM smartphones
WHERE year > 2020
ORDER BY year DESC;

SELECT * FROM temp_view_latest_phones;
```

```python
%sql
show tables;
```
